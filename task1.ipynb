```python
import requests

r = requests.get("https://gsom.spbu.ru/en/")
```

```python
print(r.status_code)
print(r.text)
```

```python
text = r.text

print("png links: ", text.count(".png"))
```

```python
end = text.find(".png") + len(".png")
beg = text.rfind("/", 0, end)
link = text[beg:end]
print(link)
```

```python
domain = "https://gsom.spbu.ru"

img = requests.get(domain + link)
with open("image.png", "wb") as file:
    file.write(img.content)
    file.close()
```
